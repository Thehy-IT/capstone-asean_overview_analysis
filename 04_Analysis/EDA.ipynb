# Exploratory Data Analysis (EDA) - ASEAN Overview Analysis
**Project:** Capstone ASEAN Overview Analysis (2015-2025)

This notebook explores the cleaned datasets generated in `02_Data/Cleaned`.

```python
import pandas as pd
import os

cleaned_dir = '../02_Data/Cleaned'
fact_df = pd.read_csv(os.path.join(cleaned_dir, 'Fact_ASEAN_Indicators.csv'))
country_df = pd.read_csv(os.path.join(cleaned_dir, 'Dim_Country.csv'))
indicator_df = pd.read_csv(os.path.join(cleaned_dir, 'Dim_Indicator.csv'))
date_df = pd.read_csv(os.path.join(cleaned_dir, 'Dim_Date.csv'))

print('Fact Indicators Shape:', fact_df.shape)
print('Dim Country Shape:', country_df.shape)
print('Dim Indicator Shape:', indicator_df.shape)
print('Dim Date Shape:', date_df.shape)
```

## Summary Statistics & Record Counts per Country

```python
merged_df = fact_df.merge(country_df, on='CountryCode').merge(indicator_df, on='SeriesCode')
print('Domain Distribution:')
print(merged_df['Domain'].value_counts())

print('\nCountry Record Counts:')
print(merged_df['CountryName'].value_counts())
```
